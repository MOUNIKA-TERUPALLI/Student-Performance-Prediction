# Student Performance Prediction Using Machine Learning

A beginner-friendly binary classification project.

```
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
import joblib

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
DATA = ROOT / 'dataset' / 'student_performance.csv'
MODEL_DIR = ROOT / 'models'
VIZ_DIR = ROOT / 'visualizations'
MODEL_DIR.mkdir(exist_ok=True)
VIZ_DIR.mkdir(exist_ok=True)
```

## 1. Load and Inspect Dataset

```
df = pd.read_csv(DATA)
df.head()
```

```
print('Shape:', df.shape)
df.info()
print('\nMissing values:\n', df.isnull().sum())
print('\nDuplicates:', df.duplicated().sum())
df.describe()
```

## 2. Exploratory Data Analysis

```
df['Result'].value_counts()
```

```
plt.figure(figsize=(7,5))
for label in ['Fail','Pass']:
    s = df[df['Result']==label]
    plt.scatter(s['StudyHours'], s['PreviousScore'], label=label, alpha=0.65)
plt.xlabel('Study Hours'); plt.ylabel('Previous Score'); plt.title('Study Hours vs Previous Score by Result'); plt.legend(); plt.tight_layout(); plt.show()
```

```
plt.figure(figsize=(7,5))
for label in ['Fail','Pass']:
    s = df[df['Result']==label]
    plt.scatter(s['Attendance'], s['PreviousScore'], label=label, alpha=0.65)
plt.xlabel('Attendance (%)'); plt.ylabel('Previous Score'); plt.title('Attendance vs Previous Score by Result'); plt.legend(); plt.tight_layout(); plt.show()
```

```
plt.figure(figsize=(7,5))
plt.hist(df['PreviousScore'], bins=12)
plt.xlabel('Previous Score'); plt.ylabel('Number of Students'); plt.title('Previous Score Distribution'); plt.tight_layout(); plt.show()
```

```
counts = df['Result'].value_counts()
plt.figure(figsize=(6,6))
plt.pie(counts.values, labels=counts.index, autopct='%1.1f%%', startangle=90)
plt.title('Student Result Distribution'); plt.show()
```

## 3. Preprocessing and Model Training

```
clean = df.drop_duplicates().copy()
clean['Result'] = clean['Result'].map({'Fail':0, 'Pass':1})
X = clean[['StudyHours','Attendance','PreviousScore']]
y = clean['Result']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)
model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
```

## 4. Model Evaluation

```
accuracy = accuracy_score(y_test, y_pred)
print(f'Accuracy: {accuracy:.2%}')
print('\nConfusion Matrix:\n', confusion_matrix(y_test, y_pred))
print('\nClassification Report:\n', classification_report(y_test, y_pred, target_names=['Fail','Pass']))
```

## 5. Save Model

```
joblib.dump(model, MODEL_DIR / 'student_model.pkl')
print('Saved:', MODEL_DIR / 'student_model.pkl')
```

## 6. New Student Prediction

```
new_student = pd.DataFrame([{'StudyHours':7, 'Attendance':88, 'PreviousScore':75}])
pred = model.predict(new_student)[0]
print('Predicted Result:', 'PASS' if pred == 1 else 'FAIL')
```
